```python
from roboflow import Roboflow
from ultralytics import YOLO
from pathlib import Path
import yaml
import shutil
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display
print("Done")
```

```python
# ── MS-PlateNet V5 | BaoVisionPlate Version 5 ──────────────────────
# Dataset : Augmentation Only (No Preprocessing)
# Models  : YOLOv8 Nano / Small / Medium

ROBOFLOW_API_KEY  = "y2yrYYGxDkzrbyEzQ4Ta"
WORKSPACE_NAME    = "muhammads-workspace-8j3io"
PROJECT_NAME      = "baovisionplate"
VERSION_NUMBER    = 5



DATASET_PATH = Path(r"C:\Users\USER 1\Desktop\BaoBao\Education\UTEM\Tahun 2 Semester 1\BITI 3413 Natural Language Processing\Notebook Jupyter\MS-PlateNet\Version-5-Dataset")

RESULTS_ROOT = Path(r"C:\Users\USER 1\Desktop\BaoBao\Education\UTEM\Tahun 2 Semester 1\BITI 3413 Natural Language Processing\Notebook Jupyter\MS-PlateNet\Version-5-runs")

CLASS_NAME   = "BaoVisionPlate"

# Training hyperparameters (identical to Version 2)
EPOCHS    = 100
IMGSZ     = 640
BATCH     = 16
PATIENCE  = 20
DEVICE    = 0
AMP       = True

DATASET_PATH.mkdir(parents=True, exist_ok=True)
RESULTS_ROOT.mkdir(parents=True, exist_ok=True)

print("Dataset path :", DATASET_PATH)
print("Results root :", RESULTS_ROOT)
print("Done")
```

```python
# Download BaoVisionPlate dataset
rf      = Roboflow(api_key=ROBOFLOW_API_KEY)
project = rf.workspace(WORKSPACE_NAME).project(PROJECT_NAME)
version = project.version(VERSION_NUMBER)
dataset = version.download("yolov8", location=str(DATASET_PATH))

print("Downloaded dataset to", dataset.location)
```

```python
# Fix data.yaml so paths are correct
yaml_path = DATASET_PATH / "data.yaml"

yaml_content = {
    "path" : str(DATASET_PATH).replace("\\", "/"),
    "train": "train/images",
    "val"  : "valid/images",
    "test" : "test/images",
    "nc"   : 1,
    "names": [CLASS_NAME],
}

with open(yaml_path, "w", encoding="utf-8") as f:
    yaml.dump(yaml_content, f, sort_keys=False, allow_unicode=True)

print("data.yaml fixed at", yaml_path)

with open(yaml_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# ── Train YOLOv8 Nano ────────────────────────────────────────────────────────
model_n   = YOLO("yolov8n.pt")

results_n = model_n.train(
    data      = str(yaml_path),
    epochs    = EPOCHS,
    imgsz     = IMGSZ,
    batch     = BATCH,
    device    = DEVICE,
    amp       = AMP,
    patience  = PATIENCE,
    project   = str(RESULTS_ROOT),
    name      = "plateNetV5nano",
    exist_ok  = True,
    pretrained= True,
    verbose   = True,
)

best_n         = RESULTS_ROOT / "plateNetV5nano" / "weights" / "best.pt"
best_n_renamed = RESULTS_ROOT / "plateNetV5nano" / "weights" / "bestNanoV5.pt"

if best_n.exists():
    shutil.copy2(best_n, best_n_renamed)
    print("Saved", best_n_renamed)
else:
    print("best.pt not found")
```

```python
# ── Train YOLOv8 Small ───────────────────────────────────────────────────────
model_s   = YOLO("yolov8s.pt")

results_s = model_s.train(
    data      = str(yaml_path),
    epochs    = EPOCHS,
    imgsz     = IMGSZ,
    batch     = BATCH,
    device    = DEVICE,
    amp       = AMP,
    patience  = PATIENCE,
    project   = str(RESULTS_ROOT),
    name      = "plateNetV5small",
    exist_ok  = True,
    pretrained= True,
    verbose   = True,
)

best_s         = RESULTS_ROOT / "plateNetV5small" / "weights" / "best.pt"
best_s_renamed = RESULTS_ROOT / "plateNetV5small" / "weights" / "bestSmallV5.pt"

if best_s.exists():
    shutil.copy2(best_s, best_s_renamed)
    print("Saved", best_s_renamed)
else:
    print("best.pt not found")
```

```python
# ── Train YOLOv8 Medium ──────────────────────────────────────────────────────
model_m   = YOLO("yolov8m.pt")

results_m = model_m.train(
    data      = str(yaml_path),
    epochs    = EPOCHS,
    imgsz     = IMGSZ,
    batch     = BATCH,
    device    = DEVICE,
    amp       = AMP,
    patience  = PATIENCE,
    project   = str(RESULTS_ROOT),
    name      = "plateNetV5medium",
    exist_ok  = True,
    pretrained= True,
    verbose   = True,
)

best_m         = RESULTS_ROOT / "plateNetV5medium" / "weights" / "best.pt"
best_m_renamed = RESULTS_ROOT / "plateNetV5medium" / "weights" / "bestMediumV5.pt"

if best_m.exists():
    shutil.copy2(best_m, best_m_renamed)
    print("Saved", best_m_renamed)
else:
    print("best.pt not found")
```

```python
# ── Results summary for this version ─────────────────────────────────────────
import pandas as pd
import matplotlib.pyplot as plt

summary_rows = []

for model_name, results_obj in [("NanoV5", results_n), ("SmallV5", results_s), ("MediumV5", results_m)]:
    try:
        metrics = results_obj.results_dict
        summary_rows.append({
            "Model"         : model_name,
            "Dataset"       : "Version 5 - Augmentation Only (No Preprocessing)",
            "Precision"     : round(metrics.get("metrics/precision(B)", 0), 4),
            "Recall"        : round(metrics.get("metrics/recall(B)",    0), 4),
            "mAP50"         : round(metrics.get("metrics/mAP50(B)",     0), 4),
            "mAP50-95"      : round(metrics.get("metrics/mAP50-95(B)",  0), 4),
        })
    except Exception as e:
        print(f"Could not extract metrics for {model_name}: {e}")

df_summary = pd.DataFrame(summary_rows)
print(df_summary.to_string(index=False))

# Save to CSV
csv_path = RESULTS_ROOT / "summary_version5.csv"
df_summary.to_csv(csv_path, index=False)
print("\nSummary saved to", csv_path)
```
